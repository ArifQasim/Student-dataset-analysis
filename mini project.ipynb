# creating mini project of student marks dataset analysis

```python
import numpy as np
```

first we have to create a dummy student dataset for marks(100 students) data

```python
dataset = np.random.randint(0, 101, size=(100, 5))
dataset
```

# Basic Analysis

now we have to get to know how many students in our dataset

```python
students = dataset.shape[0]
students
```

```python
subjects = np.array([100,100,100,100,100])
max_marks = subjects.sum().item()
max_marks
```

Total marks of each student

```python
total_marks_each_student = dataset.sum(axis = 1)
total_marks_each_student
```

Average marks of each student

```python
avg_marks_students = np.mean(dataset, axis=1)
avg_marks_students
```

now get Highest total marks of student

```python
highest_marks = total_marks_each_student.max().item()
highest_marks
```

get the Lowest total marks

```python
lowest_marks = total_marks_each_student.min().item()
lowest_marks
```

# Subject Analysis

get avg marks in each subject

```python
avg_marks_subjects = np.mean(dataset, axis=0)
avg_marks_subjects
```

get highest marks in each subjects

```python
highest_marks_each_subjects = dataset.max(axis = 0)
highest_marks_each_subjects
```

get lowest marks in each subjects

```python
lowest_marks_each_subjects = dataset.min(axis = 0)
lowest_marks_each_subjects
```

# Top Students

```python
top_students = avg_marks_students[avg_marks_students >= 80]
top_students
```

```python
top_students_ids = np.where(avg_marks_students >= 80)[0]
top_students_ids
```

# Grade Distribution

```python
persentage_of_students = (total_marks_each_student / max_marks) * 100
persentage_of_students
```

```python
grade_a_students = persentage_of_students[persentage_of_students >= 90]
print('Numbers of A Grade Students: ',grade_a_students.size)
```

```python
grade_b_students = persentage_of_students[(persentage_of_students >= 80) & (persentage_of_students <= 89)]
print('Numbers of B Grade Students: ',grade_b_students.size)
```

```python
grade_c_students = persentage_of_students[(persentage_of_students >= 70) & (persentage_of_students <= 79)]
print('Numbers of C Grade Students: ',grade_c_students.size)
```

```python
grade_d_students = persentage_of_students[(persentage_of_students >= 60) & (persentage_of_students <= 69)]
print('Numbers of D Grade Students: ',grade_d_students.size)
```

```python
grade_e_students = persentage_of_students[(persentage_of_students >= 50) & (persentage_of_students <= 59)]
print('Numbers of E Grade Students: ',grade_e_students.size)
```

```python
grade_f_students = persentage_of_students[(persentage_of_students <= 49)]
print('Numbers of F Grade Students: ',grade_f_students.size)
```
